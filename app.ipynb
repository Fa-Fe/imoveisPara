```python
import pandas as pd

df = pd.read_csv('data.csv')
```

```python
faltantes = ['Ulianópolis','Santa Luzia do Pará','Dom Eliseu','Rondon do Pará','Bom Jesus do Tocantins','Acará']
```

```python
from IPython.display import display, clear_output
import ipywidgets as widgets



faltantes = [
    'Ulianópolis',
    'Santa Luzia do Pará',
    'Dom Eliseu',
    'Rondon do Pará',
    'Bom Jesus do Tocantins',
    'Acará'
]

seletor = widgets.Dropdown(
    options=faltantes,
    value='Acará',
    description='Município:',
    style={'description_width': 'initial'}
)

display(seletor)
saida = widgets.Output()

def atualizar(change):
    with saida:
        clear_output(wait=True)

        municipio = change['new']

        display(df[df['municipio'] == municipio])

seletor.observe(atualizar, names='value')

display(saida)

# Exibe o resultado inicial
with saida:
    display(df[df['municipio'] == seletor.value])
```

```python
dfFaltantes = df[df['municipio'].isin(faltantes)]

dfFaltantes['tipo_ocupacao'].value_counts()
#total 63
```
